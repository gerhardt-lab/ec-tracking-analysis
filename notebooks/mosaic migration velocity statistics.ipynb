```python
# =========================================================
# 2. KEEP green/red CELLS AND DEFINE MOSAIC CONTEXT
# =========================================================

import os
import numpy as np
import pandas as pd

FLOW_START_TIME_IN_H = 1.0

output_dir = output_folder
os.makedirs(output_dir, exist_ok=True)

df = migration_speed_df.copy()

# clean string columns
for col in ["condition", "color", "treatment", "experimentID", "filename_key"]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()

# keep only non-depleted green cells
df = df[
    #(df["treatment"].str.upper() == "SICTRL") &
    (df["color"].str.lower() == "red")
].copy()

print("After keeping red only:", df.shape)
display(df[["condition", "color", "treatment", "experimentID"]].drop_duplicates().head(20))

# use condition directly as mosaic context
df["mosaic_group"] = df["condition"] + "_" + df["color"].astype(str).str.strip()
# Add a combined label for plotting

# biological replicate
df["replicate"] = df["experimentID"].astype(str).str.strip()

# align time so flow onset = 0
df["time_flow_h"] = df["time_in_h"] - FLOW_START_TIME_IN_H

# keep only 0-48 h after flow onset
#df = df[(df["time_flow_h"] >= 0) & (df["time_flow_h"] <= 48)].copy()

# define windows
def assign_window(t):
    if 0<= t < 10:
        return "0-10h"
    elif 10 <= t < 20:
        return "10-20h"
    elif 20 <= t < 30:
        return "20-30h"
    elif 30 <= t < 40:
        return "30-40h"
    elif 40 <= t < 50:
        return "40-50h"
    elif 50 <= t < 60:
        return "50-60h"
    else:
        return np.nan

df["time_window"] = df["time_flow_h"].apply(assign_window)
df = df.dropna(subset=["time_window"]).copy()

# explicit order
mosaic_order = [
    "mosaic_siScr_siScr_red",
    "mosaic_siScr_siScr_static_red",
    "mosaic_siCTRL_siCTRL_red",
    "mosaic_siCTRL_siCTRL_static_red"
    
]
mosaic_order = [g for g in mosaic_order if g in df["mosaic_group"].unique().tolist()]

df["mosaic_group"] = pd.Categorical(df["mosaic_group"], categories=mosaic_order, ordered=True)
df["time_window"] = pd.Categorical(df["time_window"], categories=["0-10h", "10-20h", "20-30h", "30-40h", "40-50h", "50-60h"], ordered=True)
df["replicate"] = df["replicate"].astype("category")

# summarize nuclei within replicate x mosaic_group x window
df_rep_window = (
    df.groupby(["mosaic_group", "replicate", "time_window"], as_index=False, observed=True)
      .agg(
          mean_vel=("vel_mu_per_h", "mean"),
          mean_vel_x=("vel_x_mu_per_h", "mean"),
          n_nuclei=("TRACK_ID", "nunique")
      )
)

print("df_rep_window shape:", df_rep_window.shape)
display(df_rep_window.head())

df_rep_window.to_csv(
    os.path.join(output_dir, "mosaic_red_replicate_window_summary.csv"),
    index=False
)

print("Counts per mosaic_group x time_window:")
display(pd.crosstab(df_rep_window["mosaic_group"], df_rep_window["time_window"]))
```
